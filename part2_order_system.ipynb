```python
##Task 2 — Cart Operations
```

```python
cart = []

# --- Add item to cart ---
def add_to_cart(item_name, quantity):
    # Check if item exists in menu
    if item_name not in menu:
        print(f"'{item_name}' does not exist in the menu.")
        return

    # Check if item is available
    if menu[item_name]["available"] == False:
        print(f"'{item_name}' is currently unavailable.")
        return

    # Check if item is already in cart
    for entry in cart:
        if entry["item"] == item_name:
            entry["quantity"] = entry["quantity"] + quantity
            print(f"Updated '{item_name}' quantity to {entry['quantity']}.")
            return

    # Item not in cart yet, so add it
    new_entry = {
        "item": item_name,
        "quantity": quantity,
        "price": menu[item_name]["price"]
    }
    cart.append(new_entry)
    print(f"Added '{item_name}' x{quantity} to cart.")


# --- Remove item from cart ---
def remove_from_cart(item_name):
    for entry in cart:
        if entry["item"] == item_name:
            cart.remove(entry)
            print(f"Removed '{item_name}' from cart.")
            return
    print(f"'{item_name}' is not in the cart.")


# --- Update quantity ---
def update_quantity(item_name, new_quantity):
    for entry in cart:
        if entry["item"] == item_name:
            entry["quantity"] = new_quantity
            print(f"Updated '{item_name}' quantity to {new_quantity}.")
            return
    print(f"'{item_name}' is not in the cart.")


# --- Simulate the steps ---
add_to_cart("Paneer Tikka", 2)
print("Cart:", cart)

add_to_cart("Gulab Jamun", 1)
print("Cart:", cart)

add_to_cart("Paneer Tikka", 1)   # should update quantity to 3
print("Cart:", cart)

add_to_cart("Mystery Burger", 1) # does not exist
print("Cart:", cart)

add_to_cart("Chicken Wings", 1)  # unavailable
print("Cart:", cart)

remove_from_cart("Gulab Jamun")
print("Cart:", cart)

# --- Order Summary ---
subtotal = 0
for entry in cart:
    subtotal = subtotal + (entry["quantity"] * entry["price"])

gst = round(subtotal * 0.05, 2)
total = round(subtotal + gst, 2)

print("\n========== Order Summary ==========")
for entry in cart:
    line_total = entry["quantity"] * entry["price"]
    print(f"{entry['item']:<20} x{entry['quantity']}   ₹{line_total:.2f}")
print("------------------------------------")
print(f"Subtotal:        ₹{subtotal:.2f}")
print(f"GST (5%):        ₹{gst:.2f}")
print(f"Total Payable:   ₹{total:.2f}")
print("====================================")
```

```python
##Task 3 — Inventory Tracker with Deep Copy
```

```python
import copy

# Deep copy inventory before making any changes
inventory_backup = copy.deepcopy(inventory)

# Demonstrate deep copy works: change inventory, backup should stay the same
inventory["Paneer Tikka"]["stock"] = 999
print("inventory stock     :", inventory["Paneer Tikka"]["stock"])
print("backup stock        :", inventory_backup["Paneer Tikka"]["stock"])

# Restore inventory back to original before continuing
inventory["Paneer Tikka"]["stock"] = 10

# Deduct cart quantities from inventory
for entry in cart:
    item_name = entry["item"]
    qty_needed = entry["quantity"]
    current_stock = inventory[item_name]["stock"]

    if current_stock < qty_needed:
        print(f"Warning: Not enough stock for '{item_name}'. Have {current_stock}, need {qty_needed}. Deducting only {current_stock}.")
        inventory[item_name]["stock"] = 0
    else:
        inventory[item_name]["stock"] = current_stock - qty_needed
        print(f"Deducted {qty_needed} of '{item_name}'. Remaining stock: {inventory[item_name]['stock']}")

# Print reorder alerts
print("\n--- Reorder Alerts ---")
for item_name in inventory:
    stock = inventory[item_name]["stock"]
    reorder = inventory[item_name]["reorder_level"]
    if stock <= reorder:
        print(f"Reorder Alert: {item_name} — Only {stock} unit(s) left (reorder level: {reorder})")

# Show both to confirm they are different
print("\nInventory after changes:")
for item_name in inventory:
    print(f"  {item_name}: {inventory[item_name]['stock']}")

print("\nBackup (unchanged):")
for item_name in inventory_backup:
    print(f"  {item_name}: {inventory_backup[item_name]['stock']}")
```

```python
# Revenue per day
print("Revenue per day:")
for date in sales_log:
    daily_total = 0
    for order in sales_log[date]:
        daily_total = daily_total + order["total"]
    print(f"  {date} : ₹{daily_total:.2f}")

# Best-selling day
best_day = ""
best_revenue = 0
for date in sales_log:
    daily_total = 0
    for order in sales_log[date]:
        daily_total = daily_total + order["total"]
    if daily_total > best_revenue:
        best_revenue = daily_total
        best_day = date
print(f"\nBest-selling day: {best_day} (₹{best_revenue:.2f})")

# Most ordered item
item_counts = {}
for date in sales_log:
    for order in sales_log[date]:
        for item in order["items"]:
            if item not in item_counts:
                item_counts[item] = 0
            item_counts[item] = item_counts[item] + 1

most_ordered = ""
highest_count = 0
for item in item_counts:
    if item_counts[item] > highest_count:
        highest_count = item_counts[item]
        most_ordered = item
print(f"\nMost ordered item: {most_ordered} (in {highest_count} orders)")

# Add new day
sales_log["2025-01-05"] = [
    {"order_id": 11, "items": ["Butter Chicken", "Gulab Jamun", "Garlic Naan"], "total": 490.0},
    {"order_id": 12, "items": ["Paneer Tikka", "Rasgulla"], "total": 260.0},
]

# Reprint revenue per day after adding new day
print("\nUpdated revenue per day:")
for date in sales_log:
    daily_total = 0
    for order in sales_log[date]:
        daily_total = daily_total + order["total"]
    print(f"  {date} : ₹{daily_total:.2f}")

# Updated best-selling day
best_day = ""
best_revenue = 0
for date in sales_log:
    daily_total = 0
    for order in sales_log[date]:
        daily_total = daily_total + order["total"]
    if daily_total > best_revenue:
        best_revenue = daily_total
        best_day = date
print(f"\nUpdated best-selling day: {best_day} (₹{best_revenue:.2f})")

# Numbered list of all orders
print("\nAll Orders:")
order_number = 1
for date in sales_log:
    for order in sales_log[date]:
        items_str = ", ".join(order["items"])
        print(f"{order_number}. [{date}] Order #{order['order_id']} — ₹{order['total']:.2f} — Items: {items_str}")
        order_number = order_number + 1
```
